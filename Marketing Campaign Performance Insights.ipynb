```python
import pandas as pd

url = "https://raw.githubusercontent.com/ArchanaInsights/Datasets/main/marketing_campaign.csv"
df = pd.read_csv(url)

print(df.shape)

df.head()
```

```python
# Number of rows and columns
print(df.shape)
```

```python
# Display dataset summary and non-null values
df.info()
```

```python
# Generate descriptive statistics
df.describe()
```

```python
# Number of unique Campaign_ID values
print("Number of unique Campaign_IDs:", df['Campaign_ID'].nunique())
```

```python
# Unique values in Location
print("Unique Locations:")
print(df['Location'].unique())

# Unique values in Customer_Segment
print("\nUnique Customer Segments:")
print(df['Customer_Segment'].unique())
```

```python
# Count occurrences of each Campaign_Type
print("Campaign Type Counts:")
print(df['Campaign_Type'].value_counts())

# Count occurrences of each Channel_Used
print("\nChannel Used Counts:")
print(df['Channel_Used'].value_counts())
```

```python
# 3) Exploratory Data Analysis (EDA) and Visualization
# Campaign Performance:
# a scatter plot to visualize the relationship between Acquisition_Cost and
# ROI.
import matplotlib.pyplot as plt
import seaborn as sns

# Create scatter plot
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df, x='Acquisition_Cost', y='ROI')

# Add title and labels
plt.title('Relationship Between Acquisition Cost and ROI')
plt.xlabel('Acquisition Cost')
plt.ylabel('ROI')

# Display plot
plt.show()
```

```python
# a bar chart to visualize the average Conversion_Rate for different
# Channel_Used, categorized by Campaign_Type.

import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size
plt.figure(figsize=(6, 4))

# Create bar chart
sns.barplot(
    data=df,
    x='Channel_Used',
    y='Conversion_Rate',
    hue='Campaign_Type',
    estimator='mean'
)

# Add title and labels
plt.title('Average Conversion Rate by Channel Used and Campaign Type')
plt.xlabel('Channel Used')
plt.ylabel('Average Conversion Rate')

# Rotate labels if needed
plt.xticks(rotation=45)

# Show legend and plot
plt.legend(title='Campaign Type')
plt.show()
```

```python
# the distribution of Engagement_Score across different Campaign_Type
# using a box plot.

import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size
plt.figure(figsize=(6, 4))

# Create box plot
sns.boxplot(
    data=df,
    x='Campaign_Type',
    y='Engagement_Score'
)

# Add title and labels
plt.title('Distribution of Engagement Score by Campaign Type')
plt.xlabel('Campaign Type')
plt.ylabel('Engagement Score')

# Rotate labels if needed
plt.xticks(rotation=45)

# Display plot
plt.show()
```

```python
# the average ROI by Company using a bar chart to compare the profitability
# of campaigns conducted by different companies.

import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size
plt.figure(figsize=(6, 4))

# Create bar chart
sns.barplot(
    data=df,
    x='Company',
    y='ROI',
    estimator='mean'
)

# Add title and labels
plt.title('Average ROI by Company')
plt.xlabel('Company')
plt.ylabel('Average ROI')

# Rotate company names if needed
plt.xticks(rotation=45)

# Display plot
plt.show()
```

```python
# the correlation between Engagement_Score and Conversion_Rate
# using a heatmap.

import matplotlib.pyplot as plt
import seaborn as sns

# Select the columns and calculate correlation
corr_matrix = df[['Engagement_Score', 'Conversion_Rate']].corr()

# Create heatmap
plt.figure(figsize=(4, 2))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

# Add title
plt.title('Correlation Between Engagement Score and Conversion Rate')

# Display heatmap
plt.show()
```

```python
#Customer Segmentation:
# a count plot to visualize the distribution of Target_Audience.

import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size
plt.figure(figsize=(4, 2))

# Create count plot
sns.countplot(data=df, x='Target_Audience')

# Add title and labels
plt.title('Distribution of Target Audience')
plt.xlabel('Target Audience')
plt.ylabel('Count')

# Rotate labels if needed
plt.xticks(rotation=45)

# Display plot
plt.show()
```

```python
# Customer_Segment has the highest Conversion_Rate for each
# Language using a bar chart.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate average Conversion Rate by Language and Customer Segment
avg_conversion = df.groupby(['Language', 'Customer_Segment'])['Conversion_Rate'].mean().reset_index()

# Get the Customer Segment with the highest Conversion Rate for each Language
top_segments = avg_conversion.loc[
    avg_conversion.groupby('Language')['Conversion_Rate'].idxmax()
]

# Display the results
print(top_segments)

# Create bar chart
plt.figure(figsize=(6, 4))
sns.barplot(
    data=top_segments,
    x='Language',
    y='Conversion_Rate',
    hue='Customer_Segment'
)

plt.title('Customer Segment with Highest Conversion Rate for Each Language')
plt.xlabel('Language')
plt.ylabel('Average Conversion Rate')
plt.legend(title='Customer Segment')
plt.show()
```

```python
# the distribution of Acquisition_Cost across each Customer_Segment,
# categorized by Channel_Used, using a box plot.

import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size
plt.figure(figsize=(6, 4))

# Create box plot
sns.boxplot(
    data=df,
    x='Customer_Segment',
    y='Acquisition_Cost',
    hue='Channel_Used'
)

# Add title and labels
plt.title('Acquisition Cost by Customer Segment and Channel Used')
plt.xlabel('Customer Segment')
plt.ylabel('Acquisition Cost')

# Rotate labels if needed
plt.xticks(rotation=45)

# Show legend
plt.legend(title='Channel Used', bbox_to_anchor=(1.05, 1), loc='upper left')

# Display plot
plt.tight_layout()
plt.show()
```

```python
# average Conversion_Rate by Language using a bar chart to compare
# the effectiveness of campaigns conducted in different languages.

import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size
plt.figure(figsize=(6, 4))

# Create bar chart
sns.barplot(
    data=df,
    x='Language',
    y='Conversion_Rate',
    estimator='mean'
)

# Add title and labels
plt.title('Average Conversion Rate by Language')
plt.xlabel('Language')
plt.ylabel('Average Conversion Rate')

# Rotate labels if needed
plt.xticks(rotation=45)

# Display plot
plt.show()
```

```python
# Channel Effectiveness:
#the Engagement_Score for different Channels_Used, segmented by
# Campaign_Type, using a bar chart.

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set figure size
plt.figure(figsize=(6, 4))

# Create grouped bar chart
sns.barplot(
    data=df,
    x='Channel_Used',
    y='Engagement_Score',
    hue='Campaign_Type',
    estimator=np.mean
)

# Add title and labels
plt.title('Average Engagement Score by Channel Used and Campaign Type')
plt.xlabel('Channel Used')
plt.ylabel('Average Engagement Score')

# Display legend
plt.legend(title='Campaign Type', bbox_to_anchor=(1.05, 1), loc='upper left')

# Show plot
plt.tight_layout()
plt.show()
```

```python
# the distribution of total ROI across different Channels_Used using a pie
# chart.

import matplotlib.pyplot as plt

# Calculate total ROI by channel
roi_by_channel = df.groupby('Channel_Used')['ROI'].sum()

# Create pie chart
plt.figure(figsize=(4, 4))
plt.pie(
    roi_by_channel,
    labels=roi_by_channel.index,
    autopct='%1.1f%%',
    startangle=90
)

# Add title
plt.title('Distribution of Total ROI by Channel Used')

# Display chart
plt.show()
```

```python
# a scatter plot to show the relationship between Clicks and Impressions
# for each Channel_Used.

import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size
plt.figure(figsize=(6, 4))

# Create scatter plot
sns.scatterplot(
    data=df,
    x='Impressions',
    y='Clicks',
    hue='Channel_Used',
    alpha=0.7
)

# Add title and labels
plt.title('Relationship Between Impressions and Clicks by Channel Used')
plt.xlabel('Impressions')
plt.ylabel('Clicks')

# Display legend
plt.legend(title='Channel Used', bbox_to_anchor=(1.05, 1), loc='upper left')

# Show plot
plt.tight_layout()
plt.show()
```

```python
# Time-Based Analysis:
# the distribution of Duration using a histogram.

import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size
plt.figure(figsize=(6, 4))

# Create histogram
sns.histplot(
    data=df,
    x='Duration',
    bins=20,
    kde=True
)

# Add title and labels
plt.title('Distribution of Campaign Duration')
plt.xlabel('Duration')
plt.ylabel('Frequency')

# Display plot
plt.show()
```

```python
# the overall Conversion_Rate has changed over Date for each
# Company using a line chart.
    
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert Date column to datetime format
df['Date'] = pd.to_datetime(df['Date'] , dayfirst=True)

# Calculate average Conversion Rate by Date and Company
conversion_trend = df.groupby(['Date', 'Company'])['Conversion_Rate'].mean().reset_index()

# Create line chart
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=conversion_trend,
    x='Date',
    y='Conversion_Rate',
    hue='Company',
    marker='o'
)

# Add title and labels
plt.title('Conversion Rate Trend Over Time by Company')
plt.xlabel('Date')
plt.ylabel('Average Conversion Rate')

# Display legend
plt.legend(title='Company', bbox_to_anchor=(1.05, 1), loc='upper left')

# Improve layout
plt.tight_layout()
plt.show()
```

```python
# the trend of Engagement_Score over Date with a line chart.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')

# Calculate average Engagement Score by Date
engagement_trend = df.groupby('Date')['Engagement_Score'].mean().reset_index()

# Create line chart
plt.figure(figsize=(10, 4))

sns.lineplot(
    data=engagement_trend,
    x='Date',
    y='Engagement_Score',
    marker='o'
)

# Add title and labels
plt.title('Trend of Engagement Score Over Time')
plt.xlabel('Date')
plt.ylabel('Average Engagement Score')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Geographic Analysis:
# the highest Acquisition_Cost using a bar chart.
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate average acquisition cost by location
location_cost = df.groupby('Location')['Acquisition_Cost'].mean().reset_index()

# Sort values for better visualization
location_cost = location_cost.sort_values('Acquisition_Cost', ascending=False)

# Create bar chart
plt.figure(figsize=(10, 4))

sns.barplot(
    data=location_cost,
    x='Location',
    y='Acquisition_Cost',
    palette='viridis'
)

# Add title and labels
plt.title('Average Acquisition Cost by Location')
plt.xlabel('Location')
plt.ylabel('Average Acquisition Cost')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# the Conversion_Rate by different Location, categorized by
# Target_Audience, using a bar chart.
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set figure size
plt.figure(figsize=(10, 4))

# Create bar chart
sns.barplot(
    data=df,
    x='Location',
    y='Conversion_Rate',
    hue='Target_Audience',
    estimator=np.mean
)

# Add title and labels
plt.title('Average Conversion Rate by Location and Target Audience')
plt.xlabel('Location')
plt.ylabel('Average Conversion Rate')

# Rotate x-axis labels
plt.xticks(rotation=45)

# Show legend
plt.legend(title='Target Audience', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
# the proportion of ROI by Location using a pie chart.
import matplotlib.pyplot as plt

# Calculate total ROI by Location
roi_by_location = df.groupby('Location')['ROI'].sum()

# Create pie chart
plt.figure(figsize=(4, 4))

plt.pie(
    roi_by_location,
    labels=roi_by_location.index,
    autopct='%1.1f%%',
    startangle=90
)

# Add title
plt.title('Proportion of ROI by Location')

# Make pie chart circular
plt.axis('equal')

# Display chart
plt.show()
```
